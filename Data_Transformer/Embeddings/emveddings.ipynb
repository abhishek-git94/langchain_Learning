```python
import os
from dotenv import load_dotenv

load_dotenv()  # load all the environment variables from the .env file
```

```python
# langchain-google-genai reads GOOGLE_API_KEY from the environment automatically.
# load_dotenv() has already put it in os.environ, so there is nothing to re-assign here --
# just verify it is actually present, instead of blowing up later with a confusing error.
api_key = os.getenv("GOOGLE_API_KEY")
assert api_key, (
    "GOOGLE_API_KEY is not set. Add a line like GOOGLE_API_KEY=your-key-here to the "
    ".env file at the project root, then restart the kernel. "
    "Get a key at https://aistudio.google.com/app/apikey"
)
print("GOOGLE_API_KEY loaded:", api_key[:4] + "..." + api_key[-4:])
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings


# "gemini-1.5-*" models are chat models, not embedding models -- they have no embed endpoint.
# The current Gemini embedding model is gemini-embedding-001.
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")
embeddings
```

```python
text = "This is a tutorial"
query_vector = embeddings.embed_query(text)
query_vector
```

```python
# NOTE: GeminiEmbedding is a LlamaIndex class -- langchain_google_genai has no such name.
# In LangChain you reuse GoogleGenerativeAIEmbeddings and ask for a shorter vector via
# output_dimensionality (gemini-embedding-001 supports 128..3072 via Matryoshka truncation).
embedding_1024 = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    output_dimensionality=1024,
)
len(embedding_1024.embed_query("How are you?"))
```

```python
# Embedding several documents at once
docs = [
    "LangChain is a framework for building LLM applications.",
    "Embeddings turn text into vectors of floating point numbers.",
    "Vector stores let you search those vectors by similarity.",
]
doc_vectors = embeddings.embed_documents(docs)
print(len(doc_vectors), "vectors of dimension", len(doc_vectors[0]))
```

```python
 #vector Embedding And vector storedb
from langchain_community.vectorstores import Chroma
db = Chroma.from_texts(docs, embeddings)
```

```python
query = "it will be easier for us to conduct ourselves as belli"
retrieved_docs = db.similarity_search(query, k=2)
print("Retrieved documents:")
```
