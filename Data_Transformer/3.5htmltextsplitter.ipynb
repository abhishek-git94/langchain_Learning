```python
from langchain_text_splitters import HTMLHeaderTextSplitter
html_string = """
<!DOCTYPE html>
<html>
<head>
    <title>My Page</title>
</head>
<body>

    <h1>Hello World!</h1>
    <p>Welcome to my website.</p>

</body>
</html>
"""
header_to_split = [("h1","Header 1"), ("h2", "Header 2"), ("h3", "Header 3")]
html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=header_to_split)
html_header_split = html_splitter.split_text(html_string)
html_header_split
```
